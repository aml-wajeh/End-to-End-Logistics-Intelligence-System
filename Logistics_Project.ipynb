## 🧹 Section 6 — Data Cleaning & Validation

Professional-grade cleaning follows a principled sequence: logical validation first, then imputation, then outlier removal. Each step is documented with before/after counts.

### 6.1 — Logical Timeline Validation

**Business Rules** — The following must always hold for a valid shipment record:
- `actual_eta ≥ BookingID_Date` — an arrival cannot precede the booking
- `trip_start_date ≥ BookingID_Date` — a trip cannot start before booking
- `trip_end_date ≥ trip_start_date` — trips cannot end before starting
- All dates must be ≥ year 2010 (pre-2010 records are data entry errors)

Rows violating any rule are removed as they introduce inconsistencies that would corrupt all downstream date-based features.


```python
initial_count = len(df)

r1 = df['actual_eta'].notna()       & (df['actual_eta']       < df['BookingID_Date'])
r2 = df['trip_start_date'].notna()  & (df['trip_start_date']  < df['BookingID_Date'])
r3 = (df['trip_end_date'].notna()   & df['trip_start_date'].notna() &
      (df['trip_end_date'] < df['trip_start_date']))
r4 = ((df['trip_start_date'].dt.year < 2010) |
      (df['trip_end_date'].dt.year   < 2010)).fillna(False)

invalid_mask = r1 | r2 | r3 | r4

print('=' * 55)
print('  LOGICAL TIMELINE AUDIT')
print('=' * 55)
print(f'  Initial rows             : {initial_count:,}')
print(f'  Rule 1 (arr < booking)   : {r1.sum():,}')
print(f'  Rule 2 (start < booking) : {r2.sum():,}')
print(f'  Rule 3 (end < start)     : {r3.sum():,}')
print(f'  Rule 4 (year < 2010)     : {r4.sum():,}')
print(f'  Total invalid rows       : {invalid_mask.sum():,}')
df = df[~invalid_mask].copy()
print(f'  ✅ Remaining: {len(df):,}')
```

### 6.2 — Haversine Distance Imputation

Where `TRANSPORTATION_DISTANCE_IN_KM` is missing, we derive it from the GPS coordinates using the Haversine formula (great-circle distance). This recovers missing distance values without dropping records.


```python
def haversine_distance(lat1, lon1, lat2, lon2):
    """Great-circle distance (km) between two GPS coordinates."""
    R = 6371
    phi1, phi2 = np.radians(lat1), np.radians(lat2)
    dphi       = np.radians(lat2 - lat1)
    dlambda    = np.radians(lon2 - lon1)
    a = (np.sin(dphi / 2)**2
         + np.cos(phi1) * np.cos(phi2) * np.sin(dlambda / 2)**2)
    return 2 * R * np.arctan2(np.sqrt(a), np.sqrt(1 - a))

if 'Org_lat_lon' in df.columns and 'Des_lat_lon' in df.columns:
    coord_df = pd.concat([
        df['Org_lat_lon'].str.split(',', expand=True).rename(columns={0:'org_lat',1:'org_lon'}),
        df['Des_lat_lon'].str.split(',', expand=True).rename(columns={0:'des_lat',1:'des_lon'}),
    ], axis=1).apply(pd.to_numeric, errors='coerce')

    missing_dist = df['TRANSPORTATION_DISTANCE_IN_KM'].isna()
    if missing_dist.sum() > 0:
        imputed = haversine_distance(
            coord_df.loc[missing_dist, 'org_lat'],
            coord_df.loc[missing_dist, 'org_lon'],
            coord_df.loc[missing_dist, 'des_lat'],
            coord_df.loc[missing_dist, 'des_lon'])
        df.loc[missing_dist, 'TRANSPORTATION_DISTANCE_IN_KM'] = imputed.values
        valid_imputed = imputed.notna().sum()
        print(f'  ✅ Imputed {valid_imputed:,} missing distances via Haversine formula')
    else:
        print('  ✅ No missing distances — imputation not required')
```

### 6.3 — Speed-Based Outlier Detection & Removal

We compute each shipment's average speed (km/h). Impossible values (> 120 km/h for road freight or < 1 km/h) indicate GPS tracking failures or data corruption.


```python
df['duration_hr'] = (
    (df['trip_end_date'] - df['trip_start_date']).dt.total_seconds() / 3600
)
df = df[df['duration_hr'] > 0].copy()
df['avg_speed'] = df['TRANSPORTATION_DISTANCE_IN_KM'] / df['duration_hr']

SPEED_MAX, SPEED_MIN = 120, 1
before = len(df)
df = df[(df['avg_speed'] >= SPEED_MIN) & (df['avg_speed'] <= SPEED_MAX)].copy()
print(f'  Speed filter removed {before - len(df):,} rows → {len(df):,} remaining')
print(f'  Speed range after filter: {df["avg_speed"].min():.1f} – {df["avg_speed"].max():.1f} km/h')
```

### 6.4 — Extreme Duration Filter

Trips exceeding 15 days (360 hours) almost certainly represent GPS tracking failures where the device was not turned off after delivery.


```python
MAX_DURATION_HR = 360
before = len(df)
df = df[df['duration_hr'] <= MAX_DURATION_HR].copy()
print(f'  Duration filter removed {before - len(df):,} rows → {len(df):,} remaining')
```

## ⚙️ Section 7 — Feature Engineering

Raw columns rarely capture the full operational story. We derive features that are:
- **Operationally meaningful** — directly interpretable by logistics managers
- **Statistically informative** — correlated with delay or useful for clustering
- **Available at booking time** — ensuring no data leakage in predictive models

**Features Created:**

| Feature | Source | Purpose |
|---------|--------|---------|
| `target_delay` | `actual_eta > Planned_ETA` | Binary prediction target |
| `delay_days` | ETA gap in days | Continuous delay measure |
| `pickup_hour`, `day_of_week` | `BookingID_Date` | Temporal patterns |
| `is_weekend`, `booking_month` | `BookingID_Date` | Seasonality |
| `time_period` | `pickup_hour` | Time-of-day category |
| `log_distance` | `log1p(distance)` | Normalised distance for modelling |
| `dist_bin` | Distance quartiles | Distance category for EDA |
| `origin_state`, `dest_state` | Address parsing | Geographic risk |
| `veh_cat` | Vehicle type simplification | Reduced cardinality category |
| `mat_cat` | Material description | Cargo type category |
| `route` | origin→destination | Route identifier |
| `gps_brand` | GPS provider | Technology risk category |


```python
# ── 7.1 — Target variable ────────────────────────────────────────────────────
df['target_delay'] = (df['actual_eta'] > df['Planned_ETA']).astype(int)
df['delay_days']   = (
    (df['actual_eta'] - df['Planned_ETA']).dt.total_seconds() / 86400
).round(2)

print('Target distribution (0=On-Time, 1=Delayed):')
vc = df['target_delay'].value_counts(normalize=True).mul(100).round(1)
for k, v in vc.items():
    print(f'  {"On-Time" if k==0 else "Delayed":10s} ({k}): {v}%')

# ── 7.2 — Booking time features ──────────────────────────────────────────────
df['pickup_hour']   = df['BookingID_Date'].dt.hour
df['day_of_week']   = df['BookingID_Date'].dt.dayofweek
df['booking_month'] = df['BookingID_Date'].dt.month
df['booking_year']  = df['BookingID_Date'].dt.year
df['is_weekend']    = (df['day_of_week'] >= 5).astype(int)
df['booking_period']= df['BookingID_Date'].dt.to_period('M')

time_bins   = [0, 6, 12, 17, 21, 24]
time_labels = ['Night', 'Morning', 'Afternoon', 'Evening', 'LateNight']
df['time_period'] = pd.cut(df['pickup_hour'], bins=time_bins,
                           labels=time_labels, ordered=False)

# ── 7.3 — Distance transformation ────────────────────────────────────────────
df['log_distance'] = np.log1p(df['TRANSPORTATION_DISTANCE_IN_KM'])
df['dist_bin']     = pd.cut(df['TRANSPORTATION_DISTANCE_IN_KM'],
                            bins=[0, 200, 500, 1000, 9999],
                            labels=['Short(<200)','Medium(200-500)',
                                    'Long(500-1000)','VeryLong(>1000)'])

# ── 7.4 — Location helpers ────────────────────────────────────────────────────
def extract_state(loc):
    if pd.isna(loc): return 'Unknown'
    parts = [p.strip() for p in str(loc).split(',')]
    return parts[-1] if len(parts) >= 2 else 'Unknown'

def extract_city(loc):
    if pd.isna(loc): return 'Unknown'
    parts = [p.strip() for p in str(loc).split(',')]
    return parts[-3] if len(parts) >= 3 else 'Unknown'

df['origin_state']  = df['Origin_Location'].apply(extract_state)
df['dest_state']    = df['Destination_Location'].apply(extract_state)
df['origin_city']   = df['Origin_Location'].apply(extract_city)
df['origin_prefix'] = df['OriginLocation_Code'].astype(str).str[:3].str.upper().replace('NAN','UNK')
df['dest_prefix']   = df['DestinationLocation_Code'].astype(str).str[:3].str.upper().replace('NAN','UNK')

# Route = origin_prefix → dest_prefix
df['route'] = df['origin_prefix'] + ' → ' + df['dest_prefix']

# ── 7.5 — Vehicle & material categories ──────────────────────────────────────
def veh_cat(v):
    v = str(v)
    for kw, label in [('Trailer','Trailer'),('Hcv','HCV'),('Mcv','MCV'),
                       ('Lcv','LCV'),('Container','Container')]:
        if kw in v: return label
    return 'Other'

df['veh_cat'] = df['vehicleType'].apply(veh_cat)

def mat_cat(m):
    if pd.isna(m): return 'Unknown'
    m = str(m).upper()
    if 'AUTO' in m or 'PARTS' in m:                                       return 'Auto Parts'
    if any(k in m for k in ['STARTER','ALTERNATOR','MOTOR','RECTIFIER']): return 'Electrical'
    if any(k in m for k in ['EMPTY','TRAY','BIN','FOAM','LID']):          return 'Empty/Returns'
    if 'SPARE' in m:                                                       return 'Spare Parts'
    if any(k in m for k in ['SOLENOID','ASSEMBLY','ASSY','SPINDLE']):     return 'Assemblies'
    return 'Other'

df['mat_cat']  = df['Material Shipped'].apply(mat_cat) if 'Material Shipped' in df.columns else 'Unknown'
df['gps_brand']= df['GpsProvider'].fillna('Unknown').str.split().str[0]

print(f'\n✅ Feature engineering complete.')
print(f'   Dataset: {df.shape[0]:,} rows × {df.shape[1]} columns')
```

## 🔵 Section 12 — Data Mining — K-Means Clustering

<div style="background:#EFF6FF; border-left:4px solid #2563EB; padding:14px 18px; border-radius:0 8px 8px 0; margin:10px 0;">
<b>🎯 What we are doing:</b> Applying K-Means clustering at two analytical levels — individual shipments and origin hubs — to discover naturally occurring groups in the data.<br>
<b>💡 Why it matters:</b> Segmentation reveals that the logistics network is not homogeneous. Different shipment clusters may need different SLA policies, ETA models, and operational strategies.
</div>

**Methodology:**
1. Select operational features available at booking time
2. Normalise with `StandardScaler` (K-Means is distance-sensitive)
3. Run Elbow Method + Silhouette Score to determine optimal K
4. Train final K-Means model
5. Reduce to 2D with PCA for visualisation
6. Profile each cluster to understand its operational meaning


```python
from sklearn.cluster    import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics       import silhouette_score, davies_bouldin_score

print('Clustering libraries loaded.')
```

### 12A — Shipment-Level Clustering

```python
S_FEATURES = ['TRANSPORTATION_DISTANCE_IN_KM','duration_hr','delay_days',
              'log_distance','is_weekend','pickup_hour']
S_FEATURES  = [c for c in S_FEATURES if c in df.columns]

df_s = df[S_FEATURES].dropna().copy()
print(f'  Usable rows: {len(df_s):,} / {len(df):,}')

scaler_s   = StandardScaler()
X_s_scaled = scaler_s.fit_transform(df_s)

# Elbow + Silhouette
sse, sil, dbi = [], [], []
K_range = range(2, 11)
for k in K_range:
    km  = KMeans(n_clusters=k, random_state=42, n_init=10)
    lbl = km.fit_predict(X_s_scaled)
    sse.append(km.inertia_)
    sil.append(silhouette_score(X_s_scaled, lbl, sample_size=3000, random_state=42))
    dbi.append(davies_bouldin_score(X_s_scaled, lbl))

best_k_s = K_range[int(np.argmax(sil))]
print(f'  Silhouette scores: { {k:round(s,3) for k,s in zip(K_range,sil)} }')
print(f'  Best K: {best_k_s}')

km_s    = KMeans(n_clusters=best_k_s, random_state=42, n_init=15)
df_s['Cluster'] = km_s.fit_predict(X_s_scaled)
df.loc[df_s.index, 'ShipCluster'] = df_s['Cluster']

# PCA 2D
pca2    = PCA(n_components=2, random_state=42)
X_s_pca = pca2.fit_transform(X_s_scaled)
df_s['PC1'] = X_s_pca[:, 0]
df_s['PC2'] = X_s_pca[:, 1]

ship_profile = df_s.groupby('Cluster').agg(
    Count=('TRANSPORTATION_DISTANCE_IN_KM','count'),
    AvgDistance=('TRANSPORTATION_DISTANCE_IN_KM','mean'),
    AvgDuration=('duration_hr','mean'),
    AvgDelayDays=('delay_days','mean'),
).round(2)
print('\nShipment Cluster Profiles:')
display(ship_profile)
```

```python
# ── Visualise shipment clustering ────────────────────────────────────────────
fig = plt.figure(figsize=(18, 6))
fig.suptitle('Shipment-Level K-Means Clustering', fontsize=14, fontweight='bold')
gs  = gridspec.GridSpec(1, 3, figure=fig, wspace=0.35)

ax1 = fig.add_subplot(gs[0])
ax1.plot(K_range, sse, color=PALETTE['blue'], lw=2.5, marker='o')
ax1.axvline(best_k_s, color=PALETTE['red'], lw=2, ls='--', label=f'Best K={best_k_s}')
ax1.set_title('Elbow Curve (SSE)'); ax1.set_xlabel('K'); ax1.set_ylabel('SSE')
ax1.legend(fontsize=9)

ax2 = fig.add_subplot(gs[1])
ax2.plot(K_range, sil, color=PALETTE['green'], lw=2.5, marker='s')
ax2.axvline(best_k_s, color=PALETTE['red'], lw=2, ls='--', label=f'Best K={best_k_s}')
ax2.set_title('Silhouette Score'); ax2.set_xlabel('K'); ax2.set_ylabel('Silhouette')
ax2.legend(fontsize=9)

ax3 = fig.add_subplot(gs[2])
sample_s = df_s.sample(min(2000, len(df_s)), random_state=42)
for c in range(best_k_s):
    mask = sample_s['Cluster'] == c
    ax3.scatter(sample_s.loc[mask,'PC1'], sample_s.loc[mask,'PC2'],
                color=C10[c], alpha=0.45, s=15, edgecolors='none', label=f'Cluster {c}')
cents_pca = pca2.transform(km_s.cluster_centers_)
ax3.scatter(cents_pca[:,0], cents_pca[:,1], c='black', marker='X', s=180, zorder=5,
            edgecolors='white', label='Centroids')
ax3.set_title(f'PCA Projection (K={best_k_s})'); ax3.set_xlabel('PC1'); ax3.set_ylabel('PC2')
ax3.legend(fontsize=8)

plt.tight_layout()
plt.savefig('figures/18_shipment_clustering.png', dpi=300, bbox_inches='tight')
plt.show()
print('✅ Saved: figures/18_shipment_clustering.png')
```

```python
# ── Cluster delay rate & distance boxplots ───────────────────────────────────
if 'ShipCluster' in df.columns:
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    fig.suptitle('Shipment Cluster Characteristics', fontsize=14, fontweight='bold')

    df_s_merged = df_s.copy()
    # Delay rate per cluster
    cluster_delay = df.loc[df_s.index].copy()
    cluster_delay['ShipCluster'] = df_s['Cluster'].values
    delay_rates = cluster_delay.groupby('ShipCluster')['target_delay'].mean() * 100
    axes[0].bar(delay_rates.index, delay_rates.values,
                color=[C10[i] for i in delay_rates.index], edgecolor='white')
    axes[0].set_title('Delay Rate by Shipment Cluster')
    axes[0].set_xlabel('Cluster'); axes[0].set_ylabel('Delay Rate (%)')
    for i, v in delay_rates.items():
        axes[0].text(i, v + 0.5, f'{v:.1f}%', ha='center', fontweight='bold')

    # Distance distribution per cluster
    df_s.boxplot(column='TRANSPORTATION_DISTANCE_IN_KM', by='Cluster',
                 ax=axes[1], patch_artist=True,
                 boxprops=dict(facecolor='#BFDBFE'),
                 medianprops=dict(color=PALETTE['red'], lw=2))
    axes[1].set_title('Distance Distribution by Cluster')
    axes[1].set_xlabel('Cluster'); plt.suptitle('')
    plt.tight_layout()
    plt.savefig('figures/19_cluster_profiles.png', dpi=300, bbox_inches='tight')
    plt.show()
    print('✅ Saved: figures/19_cluster_profiles.png')
```

### 12B — Hub-Level Clustering

```python
hub_agg = (df.groupby('origin_prefix').agg(
    TotalShipments=('BookingID', 'count'),
    LateCount=('target_delay', 'sum'),
    AvgDistance=('TRANSPORTATION_DISTANCE_IN_KM', 'mean'),
    AvgDuration=('duration_hr', 'mean'),
    AvgDelayDays=('delay_days', 'mean'),
    UniqueRoutes=('dest_prefix', 'nunique'),
).reset_index())
hub_agg['LateRate'] = hub_agg['LateCount'] / hub_agg['TotalShipments']
hub_agg = hub_agg[hub_agg['TotalShipments'] >= 20].dropna()
print(f'  Hubs with ≥20 shipments: {len(hub_agg)}')

H_FEATURES = ['TotalShipments','LateRate','AvgDistance','AvgDuration','AvgDelayDays','UniqueRoutes']
scaler_h   = StandardScaler()
X_h_scaled = scaler_h.fit_transform(hub_agg[H_FEATURES])

sil_h = {}
for k in range(2, min(7, len(hub_agg))):
    lbl_h = KMeans(n_clusters=k, random_state=42, n_init=10).fit_predict(X_h_scaled)
    if len(set(lbl_h)) > 1:
        sil_h[k] = silhouette_score(X_h_scaled, lbl_h)

best_k_h = max(sil_h, key=sil_h.get)
print(f'  Hub silhouette: { {k:round(v,3) for k,v in sil_h.items()} }')
print(f'  Best K: {best_k_h}')

km_h = KMeans(n_clusters=best_k_h, random_state=42, n_init=15)
hub_agg['HubCluster'] = km_h.fit_predict(X_h_scaled)

pca_h = PCA(n_components=2, random_state=42)
X_h_pca = pca_h.fit_transform(X_h_scaled)
hub_agg['PC1'] = X_h_pca[:,0]; hub_agg['PC2'] = X_h_pca[:,1]

hub_profile = hub_agg.groupby('HubCluster')[H_FEATURES].mean().round(2)
print('\nHub Cluster Profiles:')
display(hub_profile)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle('Hub-Level K-Means Clustering', fontsize=14, fontweight='bold')

for c in range(best_k_h):
    mask = hub_agg['HubCluster'] == c
    sz   = 100 + hub_agg.loc[mask,'TotalShipments'] / 30
    axes[0].scatter(hub_agg.loc[mask,'PC1'], hub_agg.loc[mask,'PC2'],
                    color=C10[c], s=sz, alpha=0.8, edgecolors='white', linewidths=1.2,
                    label=f'Cluster {c} (n={mask.sum()})')
for _, row in hub_agg.iterrows():
    axes[0].annotate(row['origin_prefix'], (row['PC1'], row['PC2']),
                     fontsize=7, ha='center', va='bottom', xytext=(0,5),
                     textcoords='offset points', color=PALETTE['gray'])
axes[0].set_title(f'Hub Clusters (K={best_k_h}) — PCA  (size ∝ volume)')
axes[0].legend(fontsize=8)

hub_agg.boxplot(column='LateRate', by='HubCluster', ax=axes[1], patch_artist=True,
                boxprops=dict(facecolor='#FCA5A5'), medianprops=dict(color=PALETTE['red'],lw=2))
axes[1].set_title('Late Rate by Hub Cluster'); axes[1].set_xlabel('Cluster')
plt.suptitle(''); plt.tight_layout()
plt.savefig('figures/20_hub_clustering.png', dpi=300, bbox_inches='tight')
plt.show()
print('✅ Saved: figures/20_hub_clustering.png')
```

## 🔗 Section 13 — Data Mining — Association Rule Mining

<div style="background:#EFF6FF; border-left:4px solid #2563EB; padding:14px 18px; border-radius:0 8px 8px 0; margin:10px 0;">
<b>🎯 What we are doing:</b> Applying the Apriori algorithm to discover frequent co-occurrence patterns across shipment attributes.<br>
<b>💡 Why it matters:</b> Association rules reveal hidden combinations of attributes that reliably predict delay — beyond what single-variable analysis can show.
</div>

**Key Metrics:**

| Metric | Definition | Business Meaning |
|--------|-----------|-----------------|
| **Support** | P(A ∩ B) | How frequently does this combination appear? |
| **Confidence** | P(B\|A) | When A appears, how often does B also appear? |
| **Lift** | Confidence / P(B) | Is this combination more frequent than chance? Lift > 1 = meaningful association |


```python
import warnings
import pandas as pd
warnings.filterwarnings('ignore')
pd.options.mode.chained_assignment = None
warnings.simplefilter(action='ignore', category=UserWarning)



from mlxtend.frequent_patterns import apriori, association_rules
from mlxtend.preprocessing    import TransactionEncoder

# ── Build transaction basket ──────────────────────────────────────────────────
# Each transaction = one shipment; items = categorical attribute:value pairs
basket_items = []
for _, row in df.iterrows():
    items = []
    if pd.notna(row.get('veh_cat')):
        items.append(f'VEH:{row["veh_cat"]}')
    if pd.notna(row.get('mat_cat')):
        items.append(f'MAT:{row["mat_cat"]}')
    if pd.notna(row.get('gps_brand')):
        items.append(f'GPS:{row["gps_brand"][:8]}')
    if pd.notna(row.get('dist_bin')):
        items.append(f'DIST:{str(row["dist_bin"])[:8]}')
    if pd.notna(row.get('time_period')):
        items.append(f'TIME:{row["time_period"]}')
    items.append('STATUS:' + ('DELAYED' if row.get('target_delay', 0) == 1 else 'ONTIME'))
    basket_items.append(items)

te  = TransactionEncoder()
te_array = te.fit_transform(basket_items)
basket_df = pd.DataFrame(te_array, columns=te.columns_)

print(f'Basket shape: {basket_df.shape[0]:,} transactions × {basket_df.shape[1]} items')
print('Sample items:', list(te.columns_)[:10])
```

```python
# ── Run Apriori ───────────────────────────────────────────────────────────────
frequent_itemsets = apriori(basket_df, min_support=0.03, use_colnames=True)
rules = association_rules(frequent_itemsets, metric='lift', min_threshold=1.05)
rules = rules.sort_values('lift', ascending=False)

print(f'Frequent itemsets: {len(frequent_itemsets):,}')
print(f'Association rules: {len(rules):,}')

# Focus on rules predicting DELAY
delay_rules = rules[rules['consequents'].astype(str).str.contains('DELAYED')]
delay_rules = delay_rules.sort_values('lift', ascending=False)

print(f'\nRules predicting DELAYED: {len(delay_rules):,}')
print('\nTop 15 rules (by Lift) predicting delay:')
display(delay_rules[['antecedents','consequents','support','confidence','lift']].head(15).round(4))
```

```python
# ── Visualise rules: support vs confidence coloured by lift ──────────────────
fig, axes = plt.subplots(1, 2, figsize=(15, 6))
fig.suptitle('Association Rule Mining — Delay Prediction Rules', fontsize=14, fontweight='bold')

if len(delay_rules) > 0:
    sc = axes[0].scatter(delay_rules['support'], delay_rules['confidence'],
                         c=delay_rules['lift'], cmap='RdYlGn_r', s=60, alpha=0.7, edgecolors='none')
    plt.colorbar(sc, ax=axes[0], label='Lift')
    axes[0].set_xlabel('Support'); axes[0].set_ylabel('Confidence')
    axes[0].set_title('Support vs Confidence (colour = Lift)')

    top10_rules = delay_rules.head(10)
    ant_labels  = [', '.join(list(a)[:2]) for a in top10_rules['antecedents']]
    axes[1].barh(range(len(top10_rules)), top10_rules['lift'].values,
                 color=PALETTE['red'], edgecolor='white', alpha=0.85)
    axes[1].set_yticks(range(len(top10_rules)))
    axes[1].set_yticklabels(ant_labels, fontsize=8)
    axes[1].set_xlabel('Lift')
    axes[1].axvline(1.0, color='black', lw=1.5, ls='--', label='Lift = 1 (baseline)')
    axes[1].set_title('Top 10 Delay Rules by Lift')
    axes[1].legend(fontsize=9)
else:
    axes[0].text(0.5, 0.5, 'No delay rules found\nTry lowering min_support',
                 ha='center', va='center', transform=axes[0].transAxes, fontsize=12)

plt.tight_layout()
plt.savefig('figures/21_association_rules.png', dpi=300, bbox_inches='tight')
plt.show()
print('✅ Saved: figures/21_association_rules.png')
```

<div style="background:#F0FDF4; border-left:4px solid #16A34A; padding:12px 18px; border-radius:0 8px 8px 0; margin:10px 0;">
<b>✅ Association Rule Insights:</b>

High-lift rules involving `STATUS:DELAYED` as the consequent reveal the most operationally significant combinations:

## 🚨 Section 14 — Data Mining — Anomaly Detection

<div style="background:#EFF6FF; border-left:4px solid #2563EB; padding:14px 18px; border-radius:0 8px 8px 0; margin:10px 0;">
<b>🎯 What we are doing:</b> Identifying structurally unusual shipments using Isolation Forest (multivariate anomaly detection).<br>
<b>💡 Why it matters:</b> Anomalies are not just outliers to remove — they are often the most important records in a logistics dataset. A shipment that took 10× longer than expected may represent a GPS failure, vehicle breakdown, customs hold, or route blockage.
</div>

**Isolation Forest** works by randomly partitioning the feature space. Anomalous points require fewer partitions to isolate (they are "isolated early"), producing a lower anomaly score.


```python
from sklearn.ensemble import IsolationForest

# Features for anomaly detection
ANOM_FEATS = [c for c in ['TRANSPORTATION_DISTANCE_IN_KM','duration_hr',
                            'avg_speed','delay_days','log_distance']
              if c in df.columns]

df_anom = df[ANOM_FEATS].dropna().copy()

iso = IsolationForest(contamination=0.05, random_state=42, n_estimators=200)

# Apply fit_predict on the feature subset
df_anom['anomaly_label']  = iso.fit_predict(df_anom[ANOM_FEATS]) # Pass only the features

# Apply score_samples on the same feature subset
df_anom['anomaly_score']  = iso.score_samples(df_anom[ANOM_FEATS]) # Pass only the features

df.loc[df_anom.index, 'anomaly_label'] = df_anom['anomaly_label']
df.loc[df_anom.index, 'anomaly_score'] = df_anom['anomaly_score']

n_anomalies = (df_anom['anomaly_label'] == -1).sum()
print(f'Anomalies detected: {n_anomalies:,} / {len(df_anom):,} ({n_anomalies/len(df_anom)*100:.1f}%)')

# Compare delay rates
df_anom['target_delay'] = df.loc[df_anom.index, 'target_delay'].values
anom_delay   = df_anom[df_anom['anomaly_label'] == -1]['target_delay'].mean()
normal_delay = df_anom[df_anom['anomaly_label'] == 1]['target_delay'].mean()
print(f'Delay rate — Anomalies  : {anom_delay*100:.1f}%')
print(f'Delay rate — Normal     : {normal_delay*100:.1f}%')
print(f'Ratio: {anom_delay/max(normal_delay,0.01):.2f}× higher in anomalies')
```

## 🤖 Section 15 — Machine Learning — Delay Classification

<div style="background:#EFF6FF; border-left:4px solid #2563EB; padding:14px 18px; border-radius:0 8px 8px 0; margin:10px 0;">
<b>🎯 What we are doing:</b> Training three classification models to predict whether a shipment will be delayed, using only information available at booking time (no future leakage).<br>
<b>💡 Why it matters:</b> A reliable delay predictor enables proactive SLA management — logistics teams can intervene for high-risk shipments before departure rather than reacting after the fact.
</div>

**Features used (all available at booking time):**
- Numerical: `log_distance`, `pickup_hour`, `day_of_week`, `is_weekend`, `booking_month`
- Categorical: `veh_cat`, `origin_state`, `dest_state`, `gps_brand`

### 15.1 — Classification Model Setup


```python
from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold
from sklearn.preprocessing  import StandardScaler, OneHotEncoder
from sklearn.impute          import SimpleImputer
from sklearn.compose         import ColumnTransformer
from sklearn.pipeline        import Pipeline
from sklearn.metrics         import (accuracy_score, precision_score, recall_score, f1_score,
                                     roc_auc_score, average_precision_score, confusion_matrix,
                                     ConfusionMatrixDisplay)
from sklearn.linear_model    import LogisticRegression
from sklearn.ensemble        import RandomForestClassifier, GradientBoostingClassifier
import joblib

print('Classification libraries loaded.')
```

### 15.2 — Feature Selection & Preprocessing Pipeline

```python
CLF_NUM_FEATS = [c for c in ['log_distance', 'pickup_hour', 'day_of_week',
                             'is_weekend', 'booking_month']
                 if c in df.columns]

CLF_CAT_FEATS = [c for c in ['veh_cat', 'origin_state', 'dest_state', 'gps_brand']
                 if c in df.columns]

TARGET = 'target_delay'
ALL_FEATS = CLF_NUM_FEATS + CLF_CAT_FEATS

df_clf = df[ALL_FEATS + [TARGET]].dropna().copy()

X = df_clf[ALL_FEATS]
y = df_clf[TARGET]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y)

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='constant', fill_value='missing')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer(transformers=[
    ('num', numeric_transformer, CLF_NUM_FEATS),
    ('cat', categorical_transformer, CLF_CAT_FEATS)
])

print(f'  Classification dataset : {len(df_clf):,} rows')
print(f'  Train / Test split     : {len(X_train):,} / {len(X_test):,}')
print(f'  Numeric features       : {CLF_NUM_FEATS}')
print(f'  Categorical features   : {CLF_CAT_FEATS}')
print(f'  Target balance         : {y.value_counts(normalize=True).round(3).to_dict()}')
```

### 15.3 — Model Training & Evaluation

```python
models = {
    'Logistic Regression': LogisticRegression(random_state=42, solver='liblinear', class_weight='balanced'),
    'Random Forest'      : RandomForestClassifier(random_state=42, n_estimators=100, class_weight='balanced'),
    'Gradient Boosting'  : GradientBoostingClassifier(random_state=42, n_estimators=100, learning_rate=0.1)
}

clf_results  = {}
best_pr_auc  = -1
best_clf_name= ''

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

for name, model in models.items():
    print(f'\n── Training {name} ─────────────────────────────────────')
    clf_pipeline = Pipeline(steps=[('preprocessor', preprocessor),
                                   ('classifier', model)])
    clf_pipeline.fit(X_train, y_train)
    y_pred  = clf_pipeline.predict(X_test)
    y_prob  = clf_pipeline.predict_proba(X_test)[:, 1]

    acc   = accuracy_score(y_test, y_pred)
    prec  = precision_score(y_test, y_pred, zero_division=0)
    rec   = recall_score(y_test, y_pred, zero_division=0)
    f1    = f1_score(y_test, y_pred, zero_division=0)
    roc   = roc_auc_score(y_test, y_prob)
    pr    = average_precision_score(y_test, y_prob)
    cv_f1 = cross_val_score(clf_pipeline, X, y, cv=cv, scoring='f1', n_jobs=-1).mean()

    clf_results[name] = {
        'Accuracy': acc, 'Precision': prec, 'Recall': rec,
        'F1 Score': f1, 'ROC-AUC': roc, 'PR-AUC': pr,
        'CV-F1': cv_f1, 'y_pred': y_pred, 'y_prob': y_prob,
        'pipeline': clf_pipeline
    }

    print(f'   Accuracy  : {acc:.4f}')
    print(f'   Precision : {prec:.4f}')
    print(f'   Recall    : {rec:.4f}')
    print(f'   F1 Score  : {f1:.4f}')
    print(f'   ROC-AUC   : {roc:.4f}')
    print(f'   PR-AUC    : {pr:.4f}  ← primary metric')
    print(f'   CV-F1 (5x): {cv_f1:.4f}')

    if pr > best_pr_auc:
        best_pr_auc  = pr
        best_clf_name= name

print(f'\n{"="*50}')
print(f'  Best model (PR-AUC): {best_clf_name} ({best_pr_auc:.4f})')
print(f'{"="*50}')
```

```python
# ── Confusion matrices ─────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle('Confusion Matrices — All Models', fontsize=14, fontweight='bold')

for ax, (name, res) in zip(axes, clf_results.items()):
    cm = confusion_matrix(y_test, res['y_pred'])
    disp = ConfusionMatrixDisplay(cm, display_labels=['On-Time', 'Delayed'])
    disp.plot(ax=ax, cmap='Blues', colorbar=False)
    ax.set_title(f'{name}\nF1={res["F1 Score"]:.3f} | PR-AUC={res["PR-AUC"]:.3f}')

plt.tight_layout()
plt.savefig('figures/23_confusion_matrices.png', dpi=300, bbox_inches='tight')
plt.show()
print('✅ Saved: figures/23_confusion_matrices.png')
```

## 📐 Section 17 — Machine Learning — Trip Duration Regression

<div style="background:#EFF6FF; border-left:4px solid #2563EB; padding:14px 18px; border-radius:0 8px 8px 0; margin:10px 0;">
<b>🎯 What we are doing:</b> Building a regression model to predict trip duration in hours — enabling better Planned ETA calibration.<br>
<b>💡 Why it matters:</b> Poorly calibrated ETAs are the root cause of SLA breaches. If we can predict actual duration more accurately, Planned ETAs can be set more realistically.
</div>


```python
try:
    from xgboost import XGBRegressor
    XGBOOST_AVAILABLE = True
    print('  XGBoost available.')
except ImportError:
    from sklearn.ensemble import GradientBoostingRegressor
    XGBOOST_AVAILABLE = False
    print('  XGBoost not available — using GradientBoostingRegressor instead.')

from sklearn.metrics import mean_absolute_error, r2_score, mean_squared_error
from sklearn.preprocessing import OrdinalEncoder

REG_FEATS = [c for c in ['log_distance','pickup_hour','day_of_week','is_weekend','booking_month']
             if c in df.columns]
REG_CAT   = [c for c in ['vehicleType','GpsProvider','veh_cat','origin_state','dest_state']
             if c in df.columns]

df_reg = df[REG_FEATS + REG_CAT + ['duration_hr']].dropna().copy()
df_reg = df_reg[df_reg['duration_hr'] < df_reg['duration_hr'].quantile(0.95)]
print(f'  Regression dataset: {len(df_reg):,} rows')

X_r = df_reg[REG_FEATS + REG_CAT]
y_r = df_reg['duration_hr']
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(
    X_r, y_r, test_size=0.25, random_state=42)

if XGBOOST_AVAILABLE:
    reg_model = XGBRegressor(n_estimators=200, max_depth=6, learning_rate=0.08,
                             random_state=42, n_jobs=-1)
else:
    reg_model = GradientBoostingRegressor(n_estimators=200, max_depth=5,
                                          learning_rate=0.08, random_state=42)

reg_pipeline = Pipeline([
    ('prep', ColumnTransformer([
        ('num', Pipeline([('impute', SimpleImputer(strategy='median'))]), REG_FEATS),
        ('cat', Pipeline([('impute', SimpleImputer(strategy='constant', fill_value='Unknown')),
                          ('encode', OrdinalEncoder(handle_unknown='use_encoded_value',
                                                    unknown_value=-1))]), REG_CAT),
    ])),
    ('reg', reg_model),
])
reg_pipeline.fit(X_train_r, y_train_r)
y_pred_r = reg_pipeline.predict(X_test_r)

mae  = mean_absolute_error(y_test_r, y_pred_r)
rmse = np.sqrt(mean_squared_error(y_test_r, y_pred_r))
r2   = r2_score(y_test_r, y_pred_r)

print(f'\n{"="*50}')
print('  REGRESSION PERFORMANCE')
print(f'{"="*50}')
print(f'  MAE  : {mae:.2f} hours  ← average prediction error')
print(f'  RMSE : {rmse:.2f} hours')
print(f'  R²   : {r2:.4f}')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('Trip Duration — Regression Evaluation', fontsize=14, fontweight='bold')

sample_idx = np.random.choice(len(y_test_r), size=min(1000, len(y_test_r)), replace=False)
axes[0].scatter(y_test_r.values[sample_idx], y_pred_r[sample_idx],
                alpha=0.35, s=15, color=PALETTE['blue'], edgecolors='none')
lim = max(y_test_r.max(), y_pred_r.max())
axes[0].plot([0, lim], [0, lim], 'r--', lw=2, label='Perfect Prediction')
axes[0].set_title(f'Actual vs Predicted Duration (R²={r2:.3f})')
axes[0].set_xlabel('Actual Duration (hr)'); axes[0].set_ylabel('Predicted Duration (hr)')
axes[0].legend()

residuals = y_test_r.values - y_pred_r
axes[1].hist(residuals, bins=50, color=PALETTE['purple'], edgecolor='white', alpha=0.85)
axes[1].axvline(0, color='black', lw=2, ls='--', label='Zero error')
axes[1].axvline(residuals.mean(), color=PALETTE['red'], lw=2,
                label=f'Mean error: {residuals.mean():.2f} hr')
axes[1].set_title('Residual Distribution')
axes[1].set_xlabel('Residual (Actual − Predicted)'); axes[1].legend()

plt.tight_layout()
plt.savefig('figures/25_regression_evaluation.png', dpi=300, bbox_inches='tight')
plt.show()
print('✅ Saved: figures/25_regression_evaluation.png')
```

### Interactive Trip Duration Predictor

```python
def predict_trip(distance_km: float, vehicle_type: str, gps_provider: str,
                 hour: int, day_of_week: int) -> None:
    is_wknd  = int(day_of_week >= 5)
    log_dist = np.log1p(distance_km)
    month    = 6

    row = {}
    for f in REG_FEATS:
        if f == 'log_distance':    row[f] = log_dist
        elif f == 'pickup_hour':   row[f] = hour
        elif f == 'day_of_week':   row[f] = day_of_week
        elif f == 'is_weekend':    row[f] = is_wknd
        elif f == 'booking_month': row[f] = month
        else:                      row[f] = np.nan

    for f in REG_CAT:
        if f == 'vehicleType':   row[f] = vehicle_type
        elif f == 'GpsProvider': row[f] = gps_provider
        elif f == 'veh_cat':     row[f] = veh_cat(vehicle_type)
        else:                    row[f] = 'Unknown'

    X_new   = pd.DataFrame([row])[REG_FEATS + REG_CAT]
    pred_hr = reg_pipeline.predict(X_new)[0]
    days, hrs = divmod(pred_hr, 24)

    print(f'\n  Shipment Duration Prediction')
    print(f'  ──────────────────────────────────────────')
    print(f'  Distance     : {distance_km:,.0f} km')
    print(f'  Vehicle      : {vehicle_type}')
    print(f'  GPS Provider : {gps_provider}')
    print(f'  Booking      : {"Sat/Sun" if is_wknd else "Weekday"}, hour {hour:02d}:00')
    print(f'  ──────────────────────────────────────────')
    print(f'  ⏱ Predicted Duration: {pred_hr:.1f} hr ({int(days)}d {hrs:.1f}h)')

# Example predictions
predict_trip(500,  'Trailer', 'Consent',  hour=9,  day_of_week=1)
predict_trip(1200, 'Hcv',     'Vamosys', hour=22, day_of_week=5)
```
